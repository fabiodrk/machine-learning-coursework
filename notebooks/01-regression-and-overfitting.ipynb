```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.metrics import mean_squared_error, r2_score

data = pd.DataFrame({
    "Country": ["Aster", "Boreal", "Cedar", "Dover", "Elm", "Fjord", "Grove", "Harbor", "Indigo", "Juniper", "Kite", "Lumen", "Mesa", "Nova", "Orchid", "Pine", "Quartz", "River", "Sol", "Terra"],
    "GDP_per_capita": [8000, 11000, 14000, 17000, 20000, 23000, 26000, 29000, 32000, 35000, 38000, 42000, 46000, 50000, 55000, 60000, 68000, 76000, 85000, 95000],
    "Life_satisfaction": [4.4, 4.7, 5.0, 5.1, 5.4, 5.7, 5.9, 6.1, 6.2, 6.4, 6.5, 6.7, 6.8, 7.0, 7.1, 7.2, 7.0, 7.5, 7.1, 7.8],
})

display(data.head())
```

```python
data.plot(kind="scatter", x="GDP_per_capita", y="Life_satisfaction", grid=True, figsize=(8, 4))
plt.title("GDP per capita and life satisfaction")
plt.show()
```

```python
X = data[["GDP_per_capita"]]
y = data["Life_satisfaction"]
linear_model = LinearRegression()
linear_model.fit(X, y)

print("Intercept:", linear_model.intercept_)
print("Coefficient:", linear_model.coef_[0])
```

```python
new_country = [[90000]]
prediction = linear_model.predict(new_country)[0]
print(f"Predicted life satisfaction: {prediction:.2f}")
```

```python
knn_model = KNeighborsRegressor(n_neighbors=3)
knn_model.fit(X, y)


comparison = pd.DataFrame({
    "GDP_per_capita": [15000, 45000, 90000],
    "Linear_prediction": linear_model.predict([[15000], [45000], [90000]]),
    "KNN_prediction": knn_model.predict([[15000], [45000], [90000]]),
})
display(comparison.round(2))
```

```python
test_gdp = np.array([[15000], [45000], [90000]])

comparison = pd.DataFrame({
    "GDP_per_capita": test_gdp.flatten(),
    "Linear_prediction": linear_model.predict(test_gdp),
})

for k in [1, 3, 7, 15]:
    knn_model = KNeighborsRegressor(n_neighbors=k)
    knn_model.fit(X, y)

    comparison[f"KNN_k={k}"] = knn_model.predict(test_gdp)

display(comparison.round(2))
```

```python
grid = np.linspace(data["GDP_per_capita"].min(), data["GDP_per_capita"].max(), 300).reshape(-1, 1)
plt.figure(figsize=(9, 5))
plt.scatter(X, y, color="black", label="observations")

for degree in [1, 2, 4, 10]:
    polynomial_model = make_pipeline(
        PolynomialFeatures(degree),
        StandardScaler(),
        LinearRegression()
        )
    polynomial_model.fit(X, y)
    plt.plot(grid, polynomial_model.predict(grid), label=f"degree {degree}")

plt.xlabel("GDP per capita")
plt.ylabel("Life satisfaction")
plt.title("Effect of polynomial degree")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42
)

rows = []
for degree in [1, 2, 3, 10]:
    model = make_pipeline(
        PolynomialFeatures(degree = degree, include_bias = False),
        StandardScaler(),
        LinearRegression()
        )
    model.fit(X_train, y_train)
    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)
    
    rows.append({
        "degree": degree,
        "train_MSE": mean_squared_error(y_train, train_pred),
        "test_MSE": mean_squared_error(y_test, test_pred)
        })


scores = pd.DataFrame(rows)
display(scores.round(3))
```
